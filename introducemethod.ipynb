```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        print(f"Hi, I'm {self.name} and I'm {self.age} years old.")
P1 = Student("Alice", 20)
P1.introduce()
```

```python
class Phone:
    def __init__(self, brand, model, storage):
        self.brand = brand
        self.model = model
        self.storage = storage

    def phone_info(self):
        print(f"This is a {self.brand} {self.model} with {self.storage}GB of storage.")

        #simulating call

    def make_call(self, number):
        print(f"Calling {number} from {self.brand} {self.model}...")
Phone1 = Phone("Apple", "iPhone 18", 1000)
Phone2 = Phone("Samsung", "Galaxy S24", 512)
Phone1.phone_info()
Phone2.phone_info()
Phone1.make_call("123-456-7890")
Phone2.make_call("098-765-4321")


```
